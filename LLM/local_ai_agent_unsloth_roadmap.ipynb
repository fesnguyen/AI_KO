# Starter Architecture Code

```python

# Minimal Local Tool Calling Agent Skeleton

import json
import requests

TOOLS = {
    "calculator": lambda x: eval(x),
}

messages = [
    {
        "role": "system",
        "content": "You are a local AI agent."
    }
]

def call_model(messages):
    response = requests.post(
        "http://localhost:11434/api/chat",
        json={
            "model": "qwen2.5",
            "messages": messages,
            "stream": False
        }
    )

    return response.json()

while True:
    user_input = input("User: ")

    messages.append({
        "role": "user",
        "content": user_input
    })

    result = call_model(messages)

    print(result)
```
